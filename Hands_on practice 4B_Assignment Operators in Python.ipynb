## Inventory Management System 

```python
stock = 150
stock -= 28
stock += 75
stock //= 12
print("Number of full Cartons : " , stock)

# Get input from the user
stock = int(input("Enter the initial stock: "))
sold = int(input("Enter the number of items sold: "))
restock = int(input("Enter the number of new items arrived: "))
carton_size = int(input("Enter the number of items per carton: "))

# Update stock after sales
stock -= sold
print("Remaining stock after sales:", stock)

# Update stock after restocking
stock += restock
print("Updated stock after restocking:", stock)

# Store the updated stock before modifying it
remaining_stock = stock

# Calculate full cartons
stock //= carton_size
print("Number of full cartons:", stock)

# Calculate remaining loose items
loose_items = remaining_stock
loose_items %= carton_size
print("Remaining loose items:", loose_items)
```
