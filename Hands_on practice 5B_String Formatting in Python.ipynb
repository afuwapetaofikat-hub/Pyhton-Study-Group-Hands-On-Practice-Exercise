## Payroll System

```python
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 7.5

gross_salary = basic_salary + bonus
tax_amount = gross_salary * (tax_rate / 100)
net_salary = gross_salary - tax_amount

print("==================================")
print("        PAYROLL SUMMARY")
print("==================================")

print("Employee :", employee_name)
print()
print("Department :", department)
print()
print("Basic Salary : N{:.2f}".format(basic_salary))
print("Bonus : N{:.2f}".format(bonus))
print("Gross Salary : N{:.2f}".format(gross_salary))
print("Tax (7.5%) : N{:.2f}".format(tax_amount))
print("Net Salary : N{:.2f}".format(net_salary))

print()
print("==================================")
```

## E-commerce Order Receipt

```python
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500
total_cost = quantity * unit_price

if quantity > 2:
    discount = total_cost * 0.10
else:
    discount = 0

final_amount = total_cost - discount

print("====================================")
print("      ONLINE STORE RECEIPT")
print("====================================")

print("Customer :", customer_name)
print()
print("Product :", product)
print()
print("Quantity :", quantity)
print()
print("Unit Price : N{:,.2f}".format(unit_price))
print()
print("Total Cost : N{:,.2f}".format(total_cost))
print()

if quantity > 2:
    print("Discount : N{:,.2f}".format(discount))
    print()
    print("Final Amount : N{:,.2f}".format(final_amount))
    print()

print("Thank you for shopping with us!")
print()
print("====================================")
```
