## Inventory Management System

```python
initial_stock = 250
sold_stock = 85
remaining_stock = initial_stock - sold_stock
print("Remaining Stock : " , remaining_stock)
reorder_needed = remaining_stock < 200
print("Reorder Needed : " , reorder_needed)
```

## Banking Withdrawal Checker

```python
account_balance = 520000
customer_withdrawal = 150000
remaining_balance = account_balance - customer_withdrawal
print("Remaining Balance : " ,remaining_balance)
withdrawal_approved = account_balance >= customer_withdrawal
print("Withdrawal Approved : " , withdrawal_approved)
```

## Smart Shopping Discount Calculator

```python
laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000
total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive
print("Total Purchase : " , total_purchase)
discount_qualified = total_purchase >= 50000
discount_amount = int(0.1 * total_purchase)
final_amount = int(total_purchase - discount_amount)
print("Discount Qualified : " , discount_qualified)
print("Discount Amount : " , discount_amount)
print("Final Amount : " , final_amount)
```
